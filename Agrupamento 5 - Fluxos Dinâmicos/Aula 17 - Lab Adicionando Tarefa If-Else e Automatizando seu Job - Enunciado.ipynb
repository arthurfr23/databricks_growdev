# Aula 17 · Lab: adicionando tarefa If/else e automatizando seu job (enunciado)

## O que é
Um desafio que junta **lógica condicional** e **execução automática** no job do projeto.

## Por que importa
Combinar decisão e automação é exatamente o que se espera de um job em produção.

## A regra de negócio
A área de crédito quer ser avisada quando a **taxa de inadimplência** da carteira passar de um limite.

O notebook `Projeto/03_gold` já calcula essa taxa e publica o task value **`taxa_inadimplencia_pct`**. O notebook `Projeto/06_alerta_risco` grava os segmentos mais arriscados em `growdev.gold.alertas_risco`.

## O desafio
No job `growdev_pipeline_emprestimos`:

1. Crie o parâmetro de job **`limite_risco_pct`** com valor `10`.
2. Crie uma tarefa **If/else** chamada **`risco_alto`**, depois da `gold`, que verifica se a taxa de inadimplência é **maior que** o limite.
3. Se for verdadeiro, rode a tarefa **`alerta_risco`** (notebook `Projeto/06_alerta_risco`, com o parâmetro `job_run_id` = `{{job.run_id}}`).
4. Garanta que **`registrar_execucao`** continue rodando sempre, também depois do novo caminho.
5. Configure o job para rodar **automaticamente** quando chegar um arquivo novo em `/Volumes/growdev/bronze/landing/loans/`.
6. Teste **os dois ramos** da condição e o **disparo automático**.

> ⏸️ **Pause o vídeo e tente agora.** Dica: a taxa da carteira fica em torno de 9%.

## Checklist de entrega
- [ ] Existe o parâmetro `limite_risco_pct`
- [ ] `risco_alto` compara `{{tasks.gold.values.taxa_inadimplencia_pct}}` com `{{job.parameters.limite_risco_pct}}`
- [ ] Com `limite_risco_pct = 10`, `alerta_risco` fica **Excluded**
- [ ] Com `limite_risco_pct = 5`, `alerta_risco` roda e grava em `growdev.gold.alertas_risco`
- [ ] `registrar_execucao` roda nos dois casos
- [ ] Um arquivo novo em `landing/loans/` dispara o job sozinho

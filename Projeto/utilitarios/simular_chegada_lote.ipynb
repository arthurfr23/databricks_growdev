# Utilitário · Simular a chegada de um novo lote de empréstimos

Copia o arquivo `dados/lote_extra/loans_lote_02.parquet` para a pasta `landing/loans/` do volume. Funciona como se o sistema de origem tivesse entregue um arquivo novo.

Use na **aula 14**: com o job configurado com **gatilho por chegada de arquivo** apontando para `/Volumes/growdev/bronze/landing/loans/`, rode este notebook e veja o job disparar sozinho.

```python
dbutils.widgets.text("catalogo", "growdev", "Catálogo")
CATALOGO = dbutils.widgets.get("catalogo")
```

```python
import os
import shutil
from datetime import datetime

origem = os.path.abspath("../../dados/lote_extra/loans_lote_02.parquet")

# Nome com data e hora: cada execução entrega um arquivo diferente
carimbo = datetime.now().strftime("%Y%m%d_%H%M%S")
destino = f"/Volumes/{CATALOGO}/bronze/landing/loans/loans_lote_{carimbo}.parquet"

shutil.copy(origem, destino)
print(f"Novo arquivo entregue: {destino}")
display(dbutils.fs.ls(f"/Volumes/{CATALOGO}/bronze/landing/loans/"))
```

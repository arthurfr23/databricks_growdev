## O que é
Pense em uma **linha de produção de carros**. Cada estação faz uma parte (chassi, motor, pintura), e alguém precisa garantir três coisas:
- **Ordem:** não dá para pintar antes de montar.
- **Horário:** a linha começa às 6h, todo dia.
- **Dependências:** se o motor atrasou, a pintura espera.

Essa coordenação tem nome: **orquestração**. No Databricks, ela faz parte do **Lakeflow**, que cobre o ciclo completo da engenharia de dados:

| Lakeflow | Papel na linha de produção |
|---|---|
| **Lakeflow Connect** | Receber a matéria-prima (ingestão de fontes externas) |
| **Lakeflow Pipelines** | Estações de transformação declarativas |
| **Lakeflow Jobs** | O gerente da linha: ordem, horário, dependências e alertas |

## Por que importa
Pipelines executados manualmente **quebram, atrasam e não escalam**: alguém esquece, roda fora de ordem ou está de férias. Orquestrar cargas de trabalho é responsabilidade central do engenheiro de dados e aparece com frequência nas vagas da área.

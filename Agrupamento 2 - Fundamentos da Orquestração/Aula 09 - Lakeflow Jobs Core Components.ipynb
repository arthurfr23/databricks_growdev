# Aula 09 · Lakeflow Jobs: componentes principais

## O que é
Comece por uma **receita de bolo**:

```
  [Separar ingredientes]
     ↓              ↓
[Bater a massa]  [Preparar a cobertura]
     ↓              ↓
       [Montar o bolo]
```

- Cada caixa é um **passo**.
- As setas dizem **o que precisa estar pronto antes**.
- Não existe seta voltando: ninguém "desbate" a massa.

Isso é um **Grafo Acíclico Dirigido (DAG)**: *grafo* porque são caixas ligadas por setas, *dirigido* porque as setas têm sentido e *acíclico* porque não há volta.

No Lakeflow Jobs:

| Componente | O que é | Na receita |
|---|---|---|
| **Job** | O fluxo completo | A receita |
| **Tarefa (task)** | Um passo do fluxo | "Bater a massa" |
| **Tipo de tarefa** | Notebook, Pipeline, SQL, Python script, Run Job, If/else, For each... | Batedeira, forno, faca |
| **Dependência** | Quem precisa terminar antes | As setas |
| **Compute** | A máquina que executa: **Serverless** (gerenciado, liga em segundos) ou **cluster de job** (criado só para a execução) | O fogão |
| **Parâmetros** | Valores que mudam a cada execução (ex.: `catalogo`) | "Bolo para 8 ou 12 pessoas" |
| **Execução (run)** | Cada vez que o job roda | Cada bolo assado |

## Por que importa
Dominar essas peças **evita configuração errada e custo desnecessário**, como deixar um cluster interativo ligado o dia todo. É o vocabulário do dia a dia dos times de dados.

## Como aplicar: explorando um job pronto
Menu lateral → **Jobs & Pipelines** → **Projeto GrowDev — Pipeline Medallion**.

> Este é o job do professor, mostrado no vídeo. Você vai criar o seu na aula 11.

1. Aba **Tasks**: identifique as 4 tarefas e as setas entre elas (o DAG).
2. Clique em uma tarefa: veja o **tipo** (Notebook), o **caminho** do notebook e o **compute** (Serverless).
3. Painel da direita: **Job parameters**, **Schedules & Triggers**, **Notifications**.
4. Aba **Runs**: abra uma execução e leia o DAG colorido (verde = sucesso, vermelho = falha).
5. Clique em uma tarefa dentro da execução para ver a **saída** do notebook.

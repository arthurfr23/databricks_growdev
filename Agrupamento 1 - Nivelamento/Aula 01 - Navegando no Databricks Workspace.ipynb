# Aula 01 · Navegando no Databricks Workspace

## O que é
Pense no Workspace como o **escritório** da equipe de dados. Nele ficam:

| Área do escritório | No Databricks | Para que serve |
|---|---|---|
| Armários e pastas | **Workspace** (pastas e notebooks) | Guardar o código |
| Arquivo de documentos | **Catalog** | Encontrar tabelas, volumes e arquivos de dados |
| Máquinas de trabalho | **Compute** | O "computador" que executa o código |
| Agenda da equipe | **Jobs & Pipelines** | Rodar tarefas sozinhas, na ordem e na hora certas |
| Mesa de consultas | **SQL Editor** | Escrever e rodar consultas SQL |

## Por que importa
Todo o módulo acontece aqui. Saber se orientar rápido no Workspace é o primeiro sinal de autonomia em um time que usa Databricks.

## Como aplicar: tour guiado
1. Menu lateral → **Workspace** → crie uma pasta com o seu nome.
2. Dentro dela, **Create → Notebook**.
3. No canto superior direito, em **Connect**, escolha **Serverless** (o compute do plano gratuito).
4. Execute as células deste notebook com **Shift + Enter**.
5. Menu lateral → **Catalog** → abra `samples` e depois o catálogo `growdev` (criado na última célula).

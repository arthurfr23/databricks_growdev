# Aula 11 · Hands-on: criando um job pela interface do Lakeflow Jobs

## O que é
A interface visual do Lakeflow Jobs permite montar um job **clicando**, sem escrever código de configuração: você escolhe os notebooks, liga as setas e define o compute.

## Por que importa
É o caminho mais rápido para tirar um pipeline do modo manual, e é a base de **todas** as configurações das próximas aulas (agendamento, If/else, retentativas...).

## Como aplicar: passo a passo
Vamos criar um job de demonstração com **duas tarefas**: bronze → silver.

### 1. Criar o job a partir de um notebook
1. Menu lateral → **Jobs & Pipelines** → **Create** → **Job**.
2. No topo, renomeie o job para `demo_primeiro_job`.
3. Primeira tarefa:
   - **Task name:** `bronze`
   - **Type:** Notebook
   - **Source:** Workspace
   - **Path:** `Projeto/01_bronze` (dentro da sua pasta do repositório)
   - **Compute:** Serverless
4. **Create task**.

### 2. Adicionar a segunda tarefa e a dependência
1. **+ Add task** → **Notebook**.
2. **Task name:** `silver` · **Path:** `Projeto/02_silver`.
3. **Depends on:** `bronze` (é isso que desenha a seta no DAG).
4. **Create task**.

### 3. Passar um parâmetro
1. Painel da direita → **Job parameters** → **Edit**.
2. Adicione `catalogo` = `growdev`.
3. Um parâmetro de job chega em **todas** as tarefas: o notebook lê com `dbutils.widgets.get("catalogo")`.

### 4. Executar e ler o resultado
1. **Run now**.
2. Aba **Runs** → clique na execução.
3. Leia o DAG: cinza = aguardando, azul = rodando, verde = sucesso, vermelho = falha.
4. Clique em `bronze` para ver a saída do notebook (os `print` e `display`).

## Erros de configuração mais comuns

| Sintoma | Causa provável | Como corrigir |
|---|---|---|
| `Notebook not found` | Caminho errado ou notebook movido | Selecione de novo pelo botão de navegação do campo **Path** |
| As tarefas rodam ao mesmo tempo | Esqueceu o **Depends on** | Edite a tarefa e escolha a dependência |
| `Table or view not found` na silver | A bronze ainda não criou a tabela | Confira a dependência e rode a bronze primeiro |
| O notebook usou `growdev` mesmo mudando o parâmetro | O nome do parâmetro não bate com o widget | Use exatamente `catalogo` |

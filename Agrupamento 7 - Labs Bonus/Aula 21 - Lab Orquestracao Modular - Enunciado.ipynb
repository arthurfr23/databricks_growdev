# Aula 21 · Lab bônus: orquestração modular (enunciado)

## O que é
Em programação, uma boa **função** faz uma coisa só e pode ser chamada por outras. Jobs modulares seguem a mesma ideia: cada job faz uma etapa, e um **job orquestrador** chama os outros na ordem certa, usando a tarefa **Run Job**.

```
growdev_orquestrador
   ├── Run Job → growdev_ingestao        (preparar_ambiente → bronze)
   └── Run Job → growdev_transformacao   (silver → gold)
```

## Por que importa
Jobs monolíticos (um job gigante que faz tudo) são difíceis de manter e testar. Com módulos:
- cada time cuida do seu job;
- a ingestão pode ser reaproveitada por vários pipelines;
- dá para rodar e testar uma etapa sozinha.

## O desafio
1. Crie o job **`growdev_ingestao`** com as tarefas `preparar_ambiente → bronze` e o parâmetro `catalogo`.
2. Crie o job **`growdev_transformacao`** com as tarefas `silver → gold` e o parâmetro `catalogo`.
3. Crie o job **`growdev_orquestrador`** com:
   - o parâmetro `catalogo` = `growdev`;
   - a tarefa `executar_ingestao` (tipo **Run Job** → `growdev_ingestao`);
   - a tarefa `executar_transformacao` (tipo **Run Job** → `growdev_transformacao`), depois da ingestão;
   - o valor de `catalogo` **repassado** do orquestrador para os dois jobs.
4. Execute **só o orquestrador** e confira que os outros dois jobs rodaram.

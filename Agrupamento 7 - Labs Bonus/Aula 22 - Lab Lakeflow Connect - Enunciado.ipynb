# Aula 22 · Lab bônus: Lakeflow Connect (enunciado)

## O que é
Pense em uma **transportadora contratada**: em vez de você mesmo ir buscar a mercadoria em cada fornecedor, ela busca, confere o que já foi entregue e traz só o que é novo.

O **Lakeflow Connect** é a camada de ingestão gerenciada do Lakeflow, com dois tipos de conector:

| Tipo | Fontes | Exemplo |
|---|---|---|
| **Conectores gerenciados** | Aplicações SaaS (Salesforce, Workday, Google Analytics...) e bancos de dados (SQL Server, MySQL, PostgreSQL) | Ingerir a tabela de empréstimos direto do MySQL do sistema de crédito |
| **Conectores padrão** | Arquivos em armazenamento na nuvem e volumes, SFTP, Kafka, upload de arquivos | Ingerir os arquivos que chegam em `landing/loans/` |

Nos dois casos o resultado é um **pipeline** que ingere de forma **incremental**: ele lembra o que já trouxe e só busca o que é novo.

## Por que importa
A ingestão costuma ser a **etapa mais frágil** de um pipeline: conexão, formato, o que já foi lido, o que mudou. Conectores gerenciados reduzem código e manutenção.

> No vídeo, o professor mostra um conector gerenciado de **MySQL**. No plano gratuito, o lab usa o conector padrão de **arquivos**, que funciona do mesmo jeito para o job.

## O desafio
1. Crie um pipeline de ingestão chamado **`growdev_ingestao_gerenciada`** que leia, de forma incremental, os arquivos Parquet de `/Volumes/growdev/bronze/landing/loans/` e grave na tabela **`growdev.bronze.loans_ingestao_gerenciada`**.
   - Use o arquivo pronto `Projeto/pipeline_ingestao/ingestao_loans.sql`.
   - **Default catalog** = `growdev` · **Default schema** = `bronze` · Compute **Serverless**.
   - Em **Configuration**, crie a chave `catalogo` = `growdev`.
2. Adicione o pipeline como **primeira etapa** do `growdev_orquestrador` (aula 21), com uma tarefa do tipo **Pipeline** chamada `ingestao_gerenciada`.
3. Rode o orquestrador de ponta a ponta.
4. Entregue um lote novo com `Projeto/utilitarios/simular_chegada_lote`, rode de novo e confira que o pipeline ingeriu **só o arquivo novo**.

# Aula 18 · Tratamento de falhas em tarefas e monitoramento do desempenho de jobs

## O que é
**Falhas vão acontecer**: a rede oscila, um arquivo chega corrompido, um serviço fica fora do ar. Pense em um **entregador**:
- Não encontrou ninguém? **Tenta de novo** mais tarde (retentativa).
- Passou do horário limite? **Desiste** e devolve ao centro (timeout).
- A entrega está demorando muito? O cliente recebe um **aviso de atraso** (alerta de duração).
- Entregou ou falhou? Manda **notificação** (início, sucesso, falha).

No Lakeflow Jobs:

| Mecanismo | Onde fica | Exemplo no projeto |
|---|---|---|
| **Retries** (retentativas) | Em cada tarefa → *Retries* | 2 tentativas, 30 s entre elas, em bronze/silver/gold |
| **Timeout** | Tarefa ou job → *Duration thresholds* | Tarefa: 30 min · Job: 60 min |
| **Alerta de duração** | Job → *Duration thresholds → Warning* | Avisar se passar de 20 min |
| **Notificações** | Job → *Job notifications* | E-mail em falha e em atraso |
| **Monitoramento** | Aba *Runs* e **Jobs & Pipelines → Job runs** | Histórico, duração, status de todos os jobs |

### Retentativa x idempotência
Retentar só é seguro se a tarefa for **idempotente**: rodar de novo não pode duplicar dado. No projeto, a bronze só ingere arquivos novos, a silver usa `MERGE` e a gold usa `overwrite`, então podemos retentar à vontade.

## Por que importa
Descobrir um problema **antes da área de negócio** e cumprir prazos de entrega de dados é o que diferencia um pipeline profissional.

## Como aplicar
### 1. Retentativas e timeout por tarefa
Para `silver` e `gold` (e para `bronze_tabela`, dentro do For each):
1. Clique na tarefa → **Retries** → **Add** → 2 tentativas, intervalo de 30 segundos.
2. **Duration threshold → Timeout** → 30 minutos.

### 2. Alerta de duração e timeout do job
Painel da direita → **Duration thresholds**:
- **Warning:** 20 minutos
- **Timeout:** 60 minutos

### 3. Notificações
Painel da direita → **Job notifications** → **Add notification**:
- Destino: o seu e-mail
- Marque **Failure** e **Duration warning**. Marque **Start** e **Success** só se fizer sentido, para não virar spam.

### 4. Lendo o painel de execuções
- No job → **Runs**: a visão em matriz mostra cada execução (colunas) e cada tarefa (linhas). Uma tarefa que sempre falha ou sempre demora aparece logo.
- **Jobs & Pipelines → Job runs**: todas as execuções do workspace, com filtros por status, job e período.
- Clique em uma execução → **Timeline**: mostra quanto tempo cada tarefa levou e o que rodou em paralelo.

## Checklist da aula
- [ ] Configurei retentativas e timeout nas tarefas
- [ ] Configurei alerta de duração e timeout do job
- [ ] Recebo e-mail quando o job falha
- [ ] Sei ler a matriz de execuções e a timeline
- [ ] Sei por que retentativa exige idempotência
